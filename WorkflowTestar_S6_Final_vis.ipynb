# PyPSA Network Visual Report

This notebook builds the comprehensive static + interactive visualization report for a PyPSA network (`.nc`) using the project module `pypsa_bc.vis.build_visual_report`.

Outputs are written to:
- `vis/vis_resources/<network_name>_plots/` (plot assets)
- `vis/<network_name>_visual_report.html` (compiled report)

```python
from pathlib import Path

from pypsa_bc.vis import build_visual_report

# Change this path if you want to report a different solved network.
NETWORK_PATH = Path("results/pypsa/PyPSA_BC_network_s_2021.nc")
VIS_DIR = Path("vis")

if not NETWORK_PATH.exists():
    raise FileNotFoundError(f"Network file not found: {NETWORK_PATH.resolve()}")

NETWORK_PATH, VIS_DIR
```

```python
result = build_visual_report(network_path=NETWORK_PATH, vis_dir=VIS_DIR)

print(f"Report: {result.report_path}")
print(f"Resources: {result.resources_dir}")
print(f"Artifacts: {len(result.artifacts)}")

result
```

```python
import pandas as pd
from IPython.display import IFrame, display

# Preview report inside notebook
if result.report_path.exists():
    display(IFrame(src=str(result.report_path), width="100%", height=720))

# Show generated assets table
artifacts_df = pd.DataFrame(
    [
        {
            "title": a.title,
            "kind": a.kind,
            "relative_path": a.relative_path,
        }
        for a in result.artifacts
    ]
)

artifacts_df
```
